```python
import pandas as pd
#Create a DataFrame
# ------------------------------------------------------------

# A DataFrame is like a table (rows and columns)

data = {
    "Product": ["Laptop", "Tablet", "Phone", "Laptop", "Phone"],
    "Price": [1000, 500, 800, 1200, 850],
    "Quantity": [5, 10, 8, 3, 6]
}

df = pd.DataFrame(data)

print("Full DataFrame:")
print(df)
```

```python
# View basic information about the data
print("First 3 rows:")
print(df.head(3))  # shows first 3 rows
print()

print("Data types and info:")
print(df.info())   # shows column types and missing values
print()

print("Summary statistics:")
print(df.describe())  # mean, std, min, max for numeric columns
print()
```

```python
# 4. Select columns
# ------------------------------------------------------------

# Select one column
print("Price column:")
print(df["Price"])
print()

# Select multiple columns
print("Product and Quantity columns:")
print(df[["Product", "Quantity"]])
print()
```

```python
# 5. Select rows
# ------------------------------------------------------------

# Select row by index number using .iloc
print("Row at index 2:")
print(df.iloc[2])
print()

# Select rows 1 to 3
print("Rows 1 to 3:")
print(df.iloc[1:4])
print()
```

```python
# 6. Filter data (Conditional selection)
# ------------------------------------------------------------

# Select rows where Price > 800
expensive_products = df[df["Price"] > 800]

print("Products with Price > 800:")
print(expensive_products)
```

```python
# 7. Add a new column
# ------------------------------------------------------------

# Revenue = Price × Quantity
df["Revenue"] = df["Price"] * df["Quantity"]

print("DataFrame with Revenue column:")
print(df)
```

```python
# 8. Calculate summary values
# ------------------------------------------------------------

print("Total Revenue:")
print(df["Revenue"].sum())
print()

print("Average Price:")
print(df["Price"].mean())
print()
```

```python
# 9. Group data
# ------------------------------------------------------------

# Group by Product and calculate total revenue per product
grouped = df.groupby("Product")["Revenue"].sum()

print("Total Revenue by Product:")
print(grouped)
print()
```

```python
# 10. Sort data
# ------------------------------------------------------------

# Sort by Price (highest first)
sorted_df = df.sort_values(by="Price", ascending=False)

print("Sorted by Price (High to Low):")
print(sorted_df)
```

```python
# 11. Save to CSV file
# ------------------------------------------------------------

# Save DataFrame to a CSV file
df.to_csv("sales_data.csv", index=False)

print("File saved as sales_data.csv")
```
