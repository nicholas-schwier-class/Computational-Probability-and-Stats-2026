## Hypergeometric distribution
Conceptual Overview

The hypergeometric distribution is used when:

We sample without replacement

From a finite population

And count the number of successes

Scenario

A box contains:

- N=20 total balls

- K=7 red balls

- 13 blue balls

We randomly select:

- n=5 balls without replacement

Find:

Probability of drawing exactly 3 red balls.

## Poisson Distribution
Conceptual Overview

The Poisson distribution models:

- The number of events occurring

- In a fixed time or space interval

- When events occur randomly and independently

- With a constant average rate

A call center receives an average of:

- λ=4 calls per hour

Find:

- Probability of exactly 2 calls in one hour

## Cumulative Distribution Function
**Conceptual Overview**

The cumulative distribution function (CDF) gives:

F(x)=P(X≤x)

It answers:

- Discrete case: Probability that a random variable is less than or equal to a value

- Continuous case: Area under the curve to the left of a value

**Scenario**

A store receives an average of:

- λ=3 customers per minute

Find:

P(X ≤ 2)

## Variance and Standard Deviation

**Conceptual Overview**

Variance measures:

- How spread out the data values are from the mean.

Standard deviation measures:

- The typical distance from the mean (in original units).

## Standard deviation represents typical distance from the mean
