Point Estimation (estimating a population value from a sample)
Real-life example

A professor wants to estimate the average study time per week of all students in a class.
She surveys only 10 students and uses their average as the estimate.

The sample mean is used as a point estimate for the population mean.

Interval Estimation (confidence intervals)
Real-life example

Instead of saying the average study time is exactly 10.1 hours, we say:

"We are 95% confident the true average study time is between 9 and 11.5 hours."

A confidence interval gives a range of plausible values for the true population mean.

Test Statistics (hypothesis testing)
Real-life example

The professor believes students study 10 hours per week on average.

We test:

H₀: μ = 10

H₁: μ ≠ 10

The test statistic tells us how far the sample mean is from the hypothesized value (in standard error units).

Linear Regression (relationship between two variables)
Real-life example

We want to see whether more study hours lead to higher exam scores.

The slope tells us how much the score increases per hour studied

R² tells us how strong the relationship is

The line is the best-fit line
