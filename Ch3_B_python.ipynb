## Hypergeometric distribution
Conceptual Overview

The hypergeometric distribution is used when:

We sample without replacement

From a finite population

And count the number of successes

Scenario

A box contains:

- N=20 total balls

- K=7 red balls

- 13 blue balls

We randomly select:

- n=5 balls without replacement

Find:

Probability of drawing exactly 3 red balls.

```python
from scipy.stats import hypergeom

# Given values
N = 20   # population size
K = 7    # number of red balls in population
n = 5    # sample size
k = 3    # number of red balls drawn

# Calculate probability
prob = hypergeom.pmf(k, N, K, n)

print("P(X = 3) =", round(prob, 4))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import hypergeom

# Parameters
N = 20
K = 7
n = 5

# Possible values of k
x = np.arange(0, n+1)

# PMF values
pmf = hypergeom.pmf(x, N, K, n)

# Plot
plt.bar(x, pmf)
plt.xlabel("Number of Red Balls Drawn")
plt.ylabel("Probability")
plt.title("Hypergeometric Distribution")
plt.show()
```

## Poisson Distribution
Conceptual Overview

The Poisson distribution models:

- The number of events occurring

- In a fixed time or space interval

- When events occur randomly and independently

- With a constant average rate

A call center receives an average of:

- λ=4 calls per hour

Find:

- Probability of exactly 2 calls in one hour

```python
from scipy.stats import poisson

# Given values
lam = 4   # average rate (lambda)
k = 2     # number of events

# Calculate probability
prob = poisson.pmf(k, lam)

print("P(X = 2) =", round(prob, 4))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lam = 4

# Reasonable range of k values
x = np.arange(0, 15)

# PMF values
pmf = poisson.pmf(x, lam)

# Plot
plt.bar(x, pmf)
plt.xlabel("Number of Calls per Hour")
plt.ylabel("Probability")
plt.title("Poisson Distribution (λ = 4)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

x = np.arange(0, 20)

pmf1 = poisson.pmf(x, 2)
pmf2 = poisson.pmf(x, 6)

plt.plot(x, pmf1, marker='o', label="λ = 2")
plt.plot(x, pmf2, marker='o', label="λ = 6")

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution Comparison")
plt.legend()
plt.show()
```

## Cumulative Distribution Function
**Conceptual Overview**

The cumulative distribution function (CDF) gives:

F(x)=P(X≤x)

It answers:

- Discrete case: Probability that a random variable is less than or equal to a value

- Continuous case: Area under the curve to the left of a value

**Scenario**

A store receives an average of:

- λ=3 customers per minute

Find:

P(X ≤ 2)

```python
from scipy.stats import poisson

# Given value
lam = 3

# CDF calculation
cdf_value = poisson.cdf(2, lam)

print("P(X ≤ 2) =", round(cdf_value, 4))
#This gives the probability of 2 or fewer customers per minute.
```

## Variance and Standard Deviation

**Conceptual Overview**

Variance measures:

- How spread out the data values are from the mean.

Standard deviation measures:

- The typical distance from the mean (in original units).

```python
import numpy as np

# Test Scores
data = np.array([70, 75, 80, 85, 90])

# Step 1: Mean
mean = np.mean(data)

# Step 2: Deviations
deviations = data - mean

# Step 3: Squared deviations
squared_dev = deviations ** 2

# Step 4: Population variance
pop_variance = np.mean(squared_dev)

# Step 5: Population standard deviation
pop_std = np.sqrt(pop_variance)

print("Mean =", mean)
print("Population Variance =", round(pop_variance, 2))
print("Population Std Dev =", round(pop_std, 2))
```

## Standard deviation represents typical distance from the mean

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([70, 75, 80, 85, 90])

mean = np.mean(data)
std = np.std(data, ddof=1)

x = np.linspace(60, 100, 100)

plt.axvline(mean)
plt.axvline(mean + std)
plt.axvline(mean - std)

plt.scatter(data, np.zeros_like(data))

plt.xlabel("Score")
plt.title("Mean and ±1 Standard Deviation")
plt.show()
```

```python
import numpy as np

data1 = np.array([70, 75, 80, 85, 90])
data2 = np.array([79, 80, 81, 80, 80])

std1 = np.std(data1, ddof=1)
std2 = np.std(data2, ddof=1)

print("Std Dev Dataset 1 =", round(std1, 3))
print("Std Dev Dataset 2 =", round(std2, 3))
```
